# Cierre de curso

Hemos aprendido lo siguiente:
- Cada Software tiene sus propias convenciones para valores faltantes, además que los valores faltantes también poseen sus propias codificaciones.
- Operaciones con valores faltantes.
- Descubrimos valores faltantes que no teníamos conciencia de su existencia: Logramos detectarlos y exponerlos para poder cuantificar su ausencia.
- Aprendimos los diferentes tipos de valores faltantes: MCAR,MAR y MNAR.
- Logramos crear visualizaciones que me ayudan a ver los patrones de estos valores faltantes, para entender si hay una relación en las variables. Entender como se acumulan y por qué suceden.
- Aprendimos las bases de la eliminación de valores faltantes, así como la imputación; es decir sustituir estos valores faltantes por algún valor estadístico, de un modelo o de un valor que asociemos por contexto.

### 
¿Cómo continuar practicando las habilidades en valores faltantes?
- Realizar el examen del curso.
- Realizar un proyecto.

Es decir realiza todo lo que hicimos en el curso con un nuevo set de datos y decide cual es lo más apropiado para el tratamiento de valores faltantes; como puede ser la eliminación de ellos o la imputación.

De tal forma que puedas realizar un análisis exploratorio con los valores faltantes y tus datos. Sin duda esto te ayudará a llevar tus habilidades al siguiente nivel.
